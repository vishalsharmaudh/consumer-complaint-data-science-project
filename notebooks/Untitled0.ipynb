```python
import pandas as pd

# Load cleaned dataset
csv_path = "/content/drive/MyDrive/Consumer Complain Project/data/processed/complaints_cleaned.csv"

df_dashboard = pd.read_csv(csv_path)

# Convert date columns to datetime
df_dashboard["date_received"] = pd.to_datetime(
    df_dashboard["date_received"],
    errors="coerce",
    utc=True
)

df_dashboard["date_sent_to_company"] = pd.to_datetime(
    df_dashboard["date_sent_to_company"],
    errors="coerce",
    utc=True
)

# Dashboard-friendly date features
df_dashboard["received_year"] = df_dashboard["date_received"].dt.year
df_dashboard["received_month"] = df_dashboard["date_received"].dt.month
df_dashboard["received_month_name"] = df_dashboard["date_received"].dt.month_name()
df_dashboard["received_quarter"] = (
    "Q" + df_dashboard["date_received"].dt.quarter.astype(str)
)
df_dashboard["received_dayofweek"] = df_dashboard["date_received"].dt.day_name()

# Response time in days
df_dashboard["response_time_days"] = (
    df_dashboard["date_sent_to_company"]
    - df_dashboard["date_received"]
).dt.total_seconds() / (24 * 60 * 60)

# Delay category for dashboard analysis
df_dashboard["response_delay_category"] = pd.cut(
    df_dashboard["response_time_days"],
    bins=[-float("inf"), 1, 3, 7, 14, 30, float("inf")],
    labels=[
        "Within 1 day",
        "2–3 days",
        "4–7 days",
        "8–14 days",
        "15–30 days",
        "Over 30 days"
    ]
)

# Save dashboard-ready dataset
dashboard_path = (
    "/content/drive/MyDrive/Consumer Complain Project/"
    "data/processed/complaints_dashboard.csv"
)

df_dashboard.to_csv(dashboard_path, index=False)

print("Dashboard dataset created successfully.")
print("Shape:", df_dashboard.shape)
print("Saved to:", dashboard_path)
print("\nNew dashboard columns:")
print([
    "received_year",
    "received_month",
    "received_month_name",
    "received_quarter",
    "received_dayofweek",
    "response_time_days",
    "response_delay_category"
])
```
